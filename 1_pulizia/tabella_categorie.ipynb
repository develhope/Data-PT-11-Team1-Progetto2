```python
import pandas as pd

df = pd.read_excel("COSMETICI.xlsx")
categorie= df["categories"]
categorie
```

```python
new_categorie = categorie.drop_duplicates().dropna().str.lower()
new_categorie
```

```python
import re


def remove_lang_prefixes(text):
    if pd.isna(text):
        return text
    # rimuove \"xx:\" all'inizio di una parola, es. \"en:soap\"
    return re.sub(r"\b[a-z]{2}:", "", str(text))

new_categorie= new_categorie.apply(remove_lang_prefixes)
new_categorie


```

```python
new_categorie.to_excel("categorie_estratte.xlsx", index=False)
from google.colab import files
files.download("categorie_estratte.xlsx")
```
