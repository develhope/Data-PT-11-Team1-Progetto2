```python
import pandas as pd
from deep_translator import GoogleTranslator


df["categories_en"] = df["categories"]  # default originale

# Prendi INDICI delle righe da tradurre
mask = (df["categories"].str.len() > 5) & df["categories"].notna()
texts_to_translate = df.loc[mask, "categories"].tolist()
indices_to_translate = df[mask].index.tolist()

print(f"Traducendo {len(texts_to_translate)} categorie...")

translator = GoogleTranslator(source='auto', target='en')

for i in range(0, len(texts_to_translate), 50):
    chunk = texts_to_translate[i:i+30]
    chunk_indices = indices_to_translate[i:i+30]

    translated = translator.translate_batch(chunk)
    df.loc[chunk_indices, "categories_en"] = translated

    print(f"Chunk {i//30+1}: {len(chunk)} ✅")

df.to_csv("categorie_EN.csv", index=False)
print("Fatto!")
print(df[["categories", "categories_en"]].head())
```
